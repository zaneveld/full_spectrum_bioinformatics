## Biology in a Nutshell 

### The BioCore survey - an overview of how to think about biology.

If you're doing something foolish, you had best try to be clever about how you go about it. Therefore, rather than trying to organize all of biology into core principles ourselves, we will start from a large scale consensus survey of experts in biology. The BioCore principles were a set of statements about biology developed from long term conversations among experts at multiple institutions ([Brownell *et al.*, 2014](https://www.lifescied.org/doi/epdf/10.1187/cbe.13-12-0233)). Those conversations did not constitute the evidence for the statements (that is embedded in thousands of papers, experiments, analyses etc), but instead merely tried to narrow down what core ideas are important for students to understand about biology.

### Disciplines and Topics

At the most coarse-grained level, the BioCore survey divided biology into three major disciplinary perspectives: Molecular, Cellular, and Developmental Biology; Physiology; and Ecology & Evolution. These represent the main perspectives that biologists use to think about and study aspects of life.  It then divided the topics that these disciplinary perspectives are used to study into some broad categories 

Let's say a little bit about each of these disciplines.

**Molecular, Cellular and Developmental Biology**. Molecular, Cellular and Developmental biology is extremely diverse, but broadly studies life from fundamental building blocks upward. This is the kind of biological approach that best matches with TV stereotypes of biologists in lab coats. Questions might include how structures within our cells form and are regulated; how compounds are moved in or out of cells; how cells sense and react to their environment; how proteins catalyze (speed up) certain chemical reactions; the exact signals that cause a heart cell to develop differently than a blood cell; how DNA replicates; etc.Importantly, all this is studied for a range of organisms, some of which are single-celled, and some of which — like viruses or prions — don't have cells at all. Some of the processes studied by molecular, cellular and developmental biology happen very fast, and as a rough trend tend to occur at faster time scales than those studied by Physiology or Ecology and Evolution.  Although there is overlap with biochemistry and organic chemistry, generally the focus is on interactions above the level of individual molecules. 

**Physiology**. Physiology is broadly concerned with how the components studied by molecular, cellular and developmental biology interact to produce function at an organismal level. Example questions might include how hormone signaling influences metabolism; how muscles produce movement; how butterflies fly; how specific mutations affect anatomy; the scaling laws relating body shape to size in different groups of organisms; etc. Studies of behavior would often be separated from physiology, but at this broad level can be included as a related discipline. It's important to recognize that physiological studies (sometimes under other names) are applied not just to mammals, but also to plants,  fungi, bacteria, etc. The time scales involved in physiological studies is generally intermediate between that of molecular biology and that of ecology and evolution.

**Ecology & Evolution**. Just as physiology thinks about how molecular processes interact to create emergent behaviors at the organismal level, ecology considers how organisms interact with each other and their environment to produce emergent outcomes at the level of entire ecosystems. In some cases the ecosystems studied may cover vast areas (e.g. if you are studying the ecology of migratory birds) or incredibly tiny (e.g. if you are studying the ecology of microbes. Stereotypically, ecology and evolution is most likely to involve extensive fieldwork studying organisms in their natural habitats. However, important ecological experiments can also happen in laboratory settings, especially for organisms like insects, slime molds, or bacteria where entire artificial ecosystems can be constructed in a small area.

Evolution studies how the traits change over time in populations or among species. This includes the study of trait change by natural selection, where the interaction between variation in traits between individuals, inheritance of traits during reproduction, and uneven survival or reproduction interact to change traits over time. It also includes neutral processes like genetic drift, gene flow between populations, and the generation of new variation through mutation. Evolutionary biology experiments and analyses try to test hypotheses about evolution. Questions might include the relationships among groups of organisms; the relationship among genes families within organisms; why certain traits evolve repeatedly in particular habitats; how different genetic variants spread or fail to spread within populations; the role of gene flow between populations in the rate of evolution; detecting gene transfer across the branches of the tree of life; or examining the signatures left by natural selection in the sequence of genes. Contrary to the stereotype that evolution is only about the past, many evolutionary studies involve active experiments, and can be very relevant for responding to modern challenges, including the evolution of new antibiotic resistance capabilities in human pathogens.

These are just some general categories. Any working biologist reading this will be ready with plenty of counter-examples that don't fit neatly into these boundaries. Indeed, much of the most exciting work in biology happens at the intersection of different disciplines. For example, molecular ecology uses molecular methods like DNA sequencing (normally the province of Molecular, Cellular and Developmental Biology) to study ecosystems. Paleontologists trying to reconstruct dinosaur communication might draw on physiology, evolution, behavior and more.

## Major Classes of Topic in Biology

Each disciplary perspective in biology can be used to try to answer many different questions. Very broadly, these questions can be divided into questions of Evolution, Structure & Function, Information Flow, Systems, and Transformations of Matter and Energy. Each of these general areas has a few overarching ideas that show up again and again, and also interacts with each of the different disciplines.

Here are some examples, quoted directly from the BioCore survey (Brownell et al., 2011):

### Evolution

    All living organisms share a common ancestor.

    Species evolve over time, and new species can arise, when allele frequencies change due to mutation, natural selection, gene flow, and genetic drift.
    
### Information Flow
    
    Organisms inherit genetic and epigenetic information that influences the location, timing, and intensity of gene expression.
    
    Cells/organs/organisms have multiple mechanisms to perceive and respond to changing environmental conditions.

### Structure
    
    Biological structures exist at all levels of organization, from molecules to ecosystems. A structure's physical and chemical characteristics influence its interactions with other structures, and therefore its function
    
    Natural selection leads to the evolution of structures that tend to increase fitness within the context of evolutionary, developmental, and environmental constraints.
    
### Transformations of Matter and Energy
    
    Energy and matter cannot be created or destroyed, but can be changed from one form to another.
    
    Energy captured by primary producers is necessary to support the maintenance, growth and reproduction of all organisms.
    
    Natural selection leads to the evolution of efficient use of resources within constraints.

## Biology basics: the central dogma

The Central Dogma describes a common pathway for information flow in cells: DNA is transcribed into RNA, which in turn can be translated into proteins that accomplish various structural and functional tasks in a cell. Let's talk through each of these stages one by one.

DNA —a long biomolecule composed of individual units called nucleotides— stores genetic information. However, that DNA generally doesn't change within an organisms lifespan, so organisms have the same DNA regardless of whether they are hot or cold, and the cells in our bodies have the same DNA regardless of whether they are a nerve cell or a liver cell. However, we know that organisms react to being hot or cold, and nerve cells and liver cells have different properties. Totally different behaviors for cells with the exact same DNA are possible because DNA does not directly carry out functions in cells. 

Instead, some select regions of the DNA are **transcribed** into temporary RNA copies. Some of those RNAs have functional roles in their own right. A few can even catalyze reactions - dramatically speeding up chemistry that is already energetically favorable. Most RNAs, however, are messenger RNAs (mRNAs) that serve as an intermediary that can bring a copy of genetic information from coding regions in the genome to ribosomes, where the genetic information in that nucleotide sequence can be **translated** into a string of amino acids that can in turn fold up into a protein.

Whether a given piece of DNA is transcribed into RNA at all, and if so *how many* RNA copies are made in a given amount of time is regulated by the interaction of signals in the DNA itself, and complex signaling networks involving the interactions between many proteins.

Only a small minority of the human genome actually encodes genes. The rest of the **non-coding sequence** contains a wild mix of sequence with different histories: randomly evolving sequence that isn't important for survival; decaying **pseudogenes** that represent the ruined  remnants of once functional genes; and **mobile genetic elements** that contain signals allowing them to be replicated in the genome even if they do not serve a function for the organism.

Much of the human genome is **non-coding sequence** that is not essential for survival. However, embedded within DNA are regulatory sequences that indicate that some sections of the DNA should be transcribed into RNA.
